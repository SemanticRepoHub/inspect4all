### Introduction to inspect4py

inspect4py is a python library or tool that parse Python code repositories to extract metadata, documentation, structure, and dependencies. It automates the analysis of

- Function, classes, and source code
- requirements (and licenses)
- repository structure (directory trees)
- control flow

The latest implementation is located at https://github.com/SemanticRepoHub/inspect4py

### Testing with inspect4py

Perform experimentation with `inspect4py` to understand how it works and the output from the tool. 

To test it with a simple python repository and a typical widely used python repository (e.g. fastapi) as samples. 

> Note The simple repository will be explored and visualised here while fastapi is explored by looking at the output directory since it is rather huge for the notebook.

The tool will be run with the following command and flags.

```sh
inspect4py -i input_directory -o output_directory -r -si -sc -ld -cl -dt -rm -md -html
```

It will generate the following:
- `-r:` requirements (dependencies)
- `-si:` software invocation
- `-sc:` source code (generate for each ast node)
- `-ld:` license detection
- `-cl:` call list
- `-dt:` directory tree (file directory tree from root path of repo)
- `-rm:` readme (extract all readme files)
- `-md:` metadata (github repository metadata using github api. input is `.git`)
- `-html:` also generate HTML output

Additionally, it also has these features that are tried out in `fastapi` sample:
- `-ast:` generate the abstract syntax tree
- `-cf:` control flow where the call graph for each file in target repository is generated
- `-df:` data flow graph (not working with fastapi)

## Misc: RepoSimPy

This section documents the experimentation with `RepoSimPy`. 

It uses the pretrained model from huggingface (Henry65/RepoSim4Py) rather than building it from scratch from its repository (https://github.com/SemanticRepoHub/RepoSim4py)

After loading the model, 2 git repository from github are "pulled" using github api (by specifying GitHub repo identifiers) to generate their embeddings vectors using the loaded model. These embedding are at various levels and a cosine similarity comparison is conducted at each level (incl. the overall level using Un
Lines 19-21: Takes two GitHub repo identifiers and generates embedding vectors for them (likely for similarity comparison)

````python
# Ensure all required dependencies (e.g. pytorch and huggingface transformers) are already installed
%pip install scikit-learn pandas torch transformers==4.46.0 
````

````python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import torch

# Verify torch is installed
import torch
print(torch.__version__)

# Verify transformers is installed
import transformers
print(transformers.__version__)

# Check if MPS or cuda is available
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"MPS available: {torch.backends.mps.is_available()}")
print(f"MPS built: {torch.backends.mps.is_built()}")

# Use GPU if available, otherwise CPU
if torch.backends.mps.is_available():
    device = "mps"
elif torch.cuda.is_available():
    device = "cuda"
else:
    device = "cpu"

print(f"Using device for inference pipeline: {device}")

# Load the RepoSim4Py model
from transformers import pipeline

model = pipeline(
    model="Henry65/RepoSim4Py", 
    trust_remote_code=True,
    device=device,
)

# Generate embeddings for the two repositories
# Here, just provide the git repo path in github. It uses github api to extract the repos

# repo1_embeddings = model("lazyhope/python-hello-world")
# repo2_embeddings = model("dbarnett/python-helloworld")
# repo1_embeddings = model("pytorch/pytorch")
# repo2_embeddings = model("tensorflow/tensorflow")
repo1_embeddings = model("fastapi/fastapi")
repo2_embeddings = model("tiangolo/full-stack-fastapi-template")

````

````python
# Extract mean embeddings for similarity computation
def extract_mean_embeddings(embeddings, keys):
    return {key: embeddings[0][key].flatten() for key in keys}

embedding_keys = [
    "mean_code_embedding",
    "mean_doc_embedding",
    "mean_readme_embedding",
    "mean_requirement_embedding",
    "mean_repo_embedding",
]

repo1_mean_embeddings = extract_mean_embeddings(repo1_embeddings, embedding_keys)
repo2_mean_embeddings = extract_mean_embeddings(repo2_embeddings, embedding_keys)

# Compute cosine similarity for each embedding level
similarity_results = {
    "code_sim": cosine_similarity(
        [repo1_mean_embeddings["mean_code_embedding"]],
        [repo2_mean_embeddings["mean_code_embedding"]],
    )[0][0],
    "doc_sim": cosine_similarity(
        [repo1_mean_embeddings["mean_doc_embedding"]],
        [repo2_mean_embeddings["mean_doc_embedding"]],
    )[0][0],
    "readme_sim": cosine_similarity(
        [repo1_mean_embeddings["mean_readme_embedding"]],
        [repo2_mean_embeddings["mean_readme_embedding"]],
    )[0][0],
    "req_sim": cosine_similarity(
        [repo1_mean_embeddings["mean_requirement_embedding"]],
        [repo2_mean_embeddings["mean_requirement_embedding"]],
    )[0][0],
    "repo_sim": cosine_similarity(
        [repo1_mean_embeddings["mean_repo_embedding"]],
        [repo2_mean_embeddings["mean_repo_embedding"]],
    )[0][0],
}

# Convert results to a DataFrame for presentation
similarity_df = pd.DataFrame([similarity_results])
similarity_df

````

As shown above, `inspect4py` extract extensive metadata information on the entire python and git based repository. The work is further extended with `RepoSim4Py` for repository similarity (using multi-level embedding and cosine similarity).
